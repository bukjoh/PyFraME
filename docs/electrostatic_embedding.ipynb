# Electrostatic Interaction

```python
from pyframe.embedding import electrostatic_interactions, particle, fragment
import qcelemental
import numpy as np
constants = qcelemental.PhysicalConstantsContext('CODATA2018')
```

## Setup system consisting of two particles

### Oxygen Atom described through a multipole expansion up to third order (charge, dipole, quadrupole, and octopole).


```python
oxygen_atom = particle.Atom(index=0, element='O',
                            mass=qcelemental.periodictable.to_mass('O'),
                            coordinate=(np.array([1.7422970, 2.3413610, -0.0007450]) / constants.bohr2angstroms),
                            multipoles={'elements': np.array([-0.7424407021, -0.2840371815, 0.1671869101,
                                                 0.0013661730, -4.4188113058, 0.2801417448,
                                                 0.0002126277, -4.1129942280, 0.0039064597,
                                                 -5.0206529493, 0.0330216712, 0.0760732530,
                                                 -0.0003899377, 0.0087745562, 0.0009840165,
                                                 -0.1813726316, -0.1218544851, -0.0016261873,
                                                 0.1033223765, 0.0025668497])},
                            polarizabilities={"elements": np.array([0., 0., 0., 0., 2.2823444229, -0.4207398269,
                                                       -0.0006457588, 1.8324399300, -0.0069062619,
                                                       3.3666855109]), "order": [1, 1]})
```

### Oxygen Nucleus

```python
oxygen_nucleus = particle.Nucleus(index=0, charge=8.0, element='O',
                                  mass=qcelemental.periodictable.to_mass('O'),
                                  coordinate=(np.array([-3.3285510, -0.1032300, -0.0004160])
                                              / constants.bohr2angstroms))
```

### Calculate potential of oxygen_atom on the coordinates of oxygen_nucleus:

```python
oxygen_atom.potential(coordinate=oxygen_nucleus.coordinate)
```

### Calculate field of oxygen_atom on the coordinates of oxygen_nucleus:

```python
oxygen_atom.potential(coordinate=oxygen_nucleus.coordinate, pot_derivative_order=1, origin_derivative_order=0)
```

### Calculate interaction energy between both particles:

```python
electrostatic_interactions.compute_atom_nucleus_interaction(oxygen_atom, oxygen_nucleus)
```
